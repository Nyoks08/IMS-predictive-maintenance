```python
# 01_eda_raw_ims.ipynb

from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

DATA_DIR = Path("../backend/data/raw/first_test")

files = sorted(DATA_DIR.iterdir())
print("Number of files:", len(files))
print("First file:", files[0].name)

x = np.loadtxt(files[0])
print("Shape:", x.shape)

plt.figure(figsize=(12, 4))
plt.plot(x[:2000, 0])
plt.title("Raw Vibration Signal (Channel 1 – first 2000 samples)")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.show()
```
