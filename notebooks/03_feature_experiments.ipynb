```python
# 03_feature_experiments.ipynb

import numpy as np
from scipy.stats import kurtosis, skew
from pathlib import Path

DATA_DIR = Path("../backend/data/raw/first_test")
x = np.loadtxt(sorted(DATA_DIR.iterdir())[0])

def extract_features(signal):
    return {
        "rms": np.sqrt(np.mean(signal**2)),
        "std": np.std(signal),
        "p2p": np.ptp(signal),
        "kurtosis": kurtosis(signal, fisher=False),
        "skew": skew(signal)
    }

features = extract_features(x[:, 0])
features
```
